# arxiv extraction

```python
import os
import gzip
import shutil
import threading
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import time
from datetime import datetime
from tqdm import tqdm
```

```python
# 配置路径
source_dir = "/dpc-zhouy/zhouy/papers/arxiv"
target_dir = "/dpc-zhouy/zhouy/papers/arxiv_demo"

# 确保目标目录存在
os.makedirs(target_dir, exist_ok=True)

# 配置线程数（根据CPU核心数调整）
MAX_WORKERS = 8  # 可以根据需要调整

print(f"源目录: {source_dir}")
print(f"目标目录: {target_dir}")
print(f"线程数: {MAX_WORKERS}")
```

```python
def decompress_file(file_path, target_dir):
    """
    解压单个文件
    """
    try:
        # 获取文件名
        file_name = os.path.basename(file_path)
        
        # 处理不同的扩展名
        if file_name.endswith('.gz.gz'):
            # 双重压缩：先解压第一层
            temp_name = file_name[:-3]  # 去掉最后一个.gz
            temp_path = os.path.join(target_dir, temp_name)
            
            # 第一层解压
            with gzip.open(file_path, 'rb') as f_in:
                with open(temp_path, 'wb') as f_out:
                    shutil.copyfileobj(f_in, f_out)
            
            # 第二层解压（如果temp_path还是.gz文件）
            if temp_name.endswith('.gz'):
                final_name = temp_name[:-3]
                final_path = os.path.join(target_dir, final_name)
                
                with gzip.open(temp_path, 'rb') as f_in:
                    with open(final_path, 'wb') as f_out:
                        shutil.copyfileobj(f_in, f_out)
                
                # 删除临时文件
                os.remove(temp_path)
                print(f"✓ 解压完成: {file_name} -> {final_name}")
            else:
                print(f"✓ 解压完成: {file_name} -> {temp_name}")
                
        elif file_name.endswith('.gz'):
            # 单层压缩
            output_name = file_name[:-3]
            output_path = os.path.join(target_dir, output_name)
            
            with gzip.open(file_path, 'rb') as f_in:
                with open(output_path, 'wb') as f_out:
                    shutil.copyfileobj(f_in, f_out)
            
            print(f"✓ 解压完成: {file_name} -> {output_name}")
        else:
            print(f"⚠ 跳过非压缩文件: {file_name}")
            
        return True, file_name
        
    except Exception as e:
        return False, f"{file_name}: {str(e)}"
```

```python
def get_files_to_decompress(source_dir):
    """
    获取所有需要解压的文件
    """
    files_to_decompress = []
    
    # 遍历源目录
    for root, dirs, files in os.walk(source_dir):
        for file in files:
            if file.endswith('.gz') or file.endswith('.gz.gz'):
                file_path = os.path.join(root, file)
                files_to_decompress.append(file_path)
    
    return files_to_decompress

# 获取文件列表
files = get_files_to_decompress(source_dir)
print(f"找到 {len(files)} 个需要解压的文件")
print(f"前5个文件: {files[:5]}")
```

```python
def single_thread_decompress(files, target_dir):
    """
    单线程解压（用于对比性能）
    """
    print("\n开始单线程解压...")
    start_time = time.time()
    
    success_count = 0
    failed_files = []
    
    for file_path in files:
        success, result = decompress_file(file_path, target_dir)
        if success:
            success_count += 1
        else:
            failed_files.append(result)
    
    elapsed_time = time.time() - start_time
    print(f"单线程解压完成: {success_count}/{len(files)} 成功")
    print(f"耗时: {elapsed_time:.2f} 秒")
    
    if failed_files:
        print(f"失败的文件: {failed_files}")
    
    return elapsed_time
```

```python
def multi_thread_decompress(files, target_dir, max_workers=8):
    """
    多线程解压
    """
    print(f"\n开始多线程解压 (线程数: {max_workers})...")
    start_time = time.time()
    
    success_count = 0
    failed_files = []
    
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        # 提交所有任务
        future_to_file = {
            executor.submit(decompress_file, file_path, target_dir): file_path 
            for file_path in files
        }
        
        # 处理完成的任务
        for future in as_completed(future_to_file):
            success, result = future.result()
            if success:
                success_count += 1
            else:
                failed_files.append(result)
    
    elapsed_time = time.time() - start_time
    print(f"多线程解压完成: {success_count}/{len(files)} 成功")
    print(f"耗时: {elapsed_time:.2f} 秒")
    
    if failed_files:
        print(f"失败的文件: {failed_files}")
    
    return elapsed_time
```

```python
# 选择解压方式
print("="*50)
print("选择解压方式:")
print("1. 单线程解压 (基准测试)")
print("2. 多线程解压 (推荐)")
print("3. 对比两种方式")
choice = input("请输入选择 (1/2/3): ")
# choice = 3

if choice == "1":
    single_time = single_thread_decompress(files[:2], target_dir)
    
elif choice == "2":
    multi_time = multi_thread_decompress(files[:2], target_dir, MAX_WORKERS)
    
elif choice == "3":
    # 先执行单线程
    single_time = single_thread_decompress(files[:2], target_dir)
    
    # 清空目标目录重新测试
    print("\n清空目标目录重新测试...")
    for item in os.listdir(target_dir):
        item_path = os.path.join(target_dir, item)
        if os.path.isfile(item_path):
            os.remove(item_path)
    
    # 再执行多线程
    multi_time = multi_thread_decompress(files[:2], target_dir, MAX_WORKERS)
    
    # 对比结果
    print("\n" + "="*50)
    print("性能对比结果:")
    print(f"单线程耗时: {single_time:.2f} 秒")
    print(f"多线程耗时: {multi_time:.2f} 秒")
    speedup = single_time / multi_time if multi_time > 0 else 0
    print(f"加速比: {speedup:.2f}x")
else:
    print("无效选择，默认执行多线程解压")
    multi_time = multi_thread_decompress(files, target_dir, MAX_WORKERS)
```

```python
def verify_extraction(target_dir):
    """
    验证解压结果
    """
    print("\n" + "="*50)
    print("验证解压结果:")
    
    extracted_files = []
    for root, dirs, files in os.walk(target_dir):
        for file in files:
            extracted_files.append(os.path.join(root, file))
    
    print(f"目标目录包含 {len(extracted_files)} 个文件")
    
    # 显示部分文件
    if extracted_files:
        print("\n前2个解压的文件:")
        for i, file_path in enumerate(extracted_files[:2]):
            file_size = os.path.getsize(file_path) / 1024. / 1024. / 1024.   # GB
            print(f"  {i+1}. {os.path.basename(file_path)} ({file_size:.2f} GB)")

        print("\n文件内容示例:")
        with open(extracted_files[0], 'r', encoding='utf-8') as f:
            lines = []
            for i in range(5):
                line = f.readline()
                print(line)
                lines.append(line.rstrip('\n'))
    return line

line_eg = verify_extraction(target_dir)
```

```python
# 统计文件夹内所有JSON文件的总行数
import os
import json
from pathlib import Path

def count_json_lines(folder_path, verbose=True):
    """
    统计文件夹内所有JSON文件的总行数
    
    参数:
        folder_path: 文件夹路径
        verbose: 是否显示详细信息
    
    返回:
        dict: 包含统计信息的字典
    """
    folder_path = Path(folder_path)
    
    if not folder_path.exists():
        raise FileNotFoundError(f"文件夹不存在: {folder_path}")
    
    if not folder_path.is_dir():
        raise NotADirectoryError(f"路径不是文件夹: {folder_path}")
    
    # 获取所有JSON文件
    json_files = list(folder_path.glob('**/*.json'))
    
    if not json_files:
        print(f"在 {folder_path} 中没有找到JSON文件")
        return {
            'total_files': 0,
            'total_lines': 0,
            'file_details': [],
            'error_files': []
        }
    
    total_lines = 0
    file_details = []
    error_files = []
    
    print(f"找到 {len(json_files)} 个JSON文件")
    print("-" * 50)
    
    for json_file in json_files:
        try:
            # 统计文件行数
            with open(json_file, 'r', encoding='utf-8') as f:
                lines = f.readlines()
                line_count = len(lines)
                total_lines += line_count
                
            file_info = {
                'file_name': json_file.name,
                'file_path': str(json_file),
                'line_count': line_count
            }
            file_details.append(file_info)
            
            if verbose:
                print(f"📄 {json_file.name}: {line_count} 行")
                
        except Exception as e:
            error_msg = f"❌ 读取文件出错 {json_file.name}: {str(e)}"
            print(error_msg)
            error_files.append({
                'file_name': json_file.name,
                'file_path': str(json_file),
                'error': str(e)
            })
    
    # 打印汇总信息
    print("-" * 50)
    print(f"📊 统计完成!")
    print(f"   文件总数: {len(json_files)}")
    print(f"   成功读取: {len(file_details)}")
    print(f"   读取失败: {len(error_files)}")
    print(f"   总行数: {total_lines:,}")
    
    if error_files:
        print(f"\n⚠️  读取失败的文件:")
        for error_file in error_files:
            print(f"   - {error_file['file_name']}: {error_file['error']}")
    
    return {
        'total_files': len(json_files),
        'successful_files': len(file_details),
        'error_files': error_files,
        'total_lines': total_lines,
        'file_details': file_details
    }

# 使用示例 - 请替换为您的文件夹路径
folder_path = '/dpc-zhouy/zhouy/papers/core-EN'  # 修改为您的JSON文件夹路径
result = count_json_lines(folder_path)
```
